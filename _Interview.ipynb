### **Character Frequency**

```python
# input = "Hello World"
# output = {'h': 1, 'e':1, 'l': 3, 'o': 2, 'w': 1, 'r': 1, 'd': 1}
mystr = "Hello World"
newstr= ''
for each in mystr:
    if each.isalpha():
        newstr = newstr + each.lower()
print(newstr)

mydict = {}
for each_1 in newstr:
    if each_1 not in mydict.keys():
        mydict[each_1] = 1
    else:
        mydict[each_1] += 1
print(mydict)
```

```python
def char_frequency(mystr):
    newstr= ''
    mydict = {}
    # Filter
    for each in mystr:
        if each.isalnum():
            newstr = newstr + each.lower()
    
    # Character frequency
    for each_1 in newstr:
        if each_1 not in mydict.keys():
            mydict[each_1] = 1
        else:
            mydict[each_1] += 1
    return mydict

ans = char_frequency("Hello World")
print(ans)
print(char_frequency("racecar"))
print(char_frequency("MaLaYalam"))
print(char_frequency("9867378547"))
```
