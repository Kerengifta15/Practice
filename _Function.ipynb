## **Python Function - user defined function**

Syntax

**def\<function_name\>():**

```python
def happy():
    print("I'm happy")
    print("code reusability")
happy()
```

```python
def happy():
    print("I'm happy")
    print("code reusability")

def sad():
    print("I'm sad")
    print("code reusability")

happy()   
sad()
```

## **func with parameter**

```python
def myfunc():
    x = 300
    y = 100
    z = x+y
    print(z)
myfunc()
```

```python
def myfunc(x,y):  #x,y are parameters
    print("x:",x)
    print("y:",y)
    print("sum:",x+y)

myfunc(10,20)  #10,20 are arguments
```

## **func with default parameter**

```python
def myfunc(x=0,y=0):  #x,y are parameters
    print("x:",x)
    print("y:",y)
    print("sum:",x+y)
myfunc()  #default values will be used
print()
myfunc(10,20)  
print()
myfunc(100)
print()
myfunc(1000)
```

### **return**

```python
def myfunc(x=0,y=0):
    print("X =",x)
    print("Y =",y)
    print("The sum of:", x+y)

ans =myfunc(100,200)
print(ans)
```

```python
def myfunc(x=0,y=0):
    print("X =",x)
    print("Y =",y)
    return x+y
myfunc(100,200)
```

## **Arbitrary Arguments**
* *args - argument stores in tuple
* **kwargs - keyword argument stores in dictionary

```python
def myfunc(x):    
    print(x,type(x))

myfunc(10)
myfunc("hai")
myfunc(True)
myfunc(50.55)
    
```

```python
def myfunc(*x):    # *args - argument stores in tuple
    print(x,type(x))
myfunc(10, "hai", True, 50.55)
```

```python
def myfunc(x):
    print(x, type(x))

myfunc(a = 10)  # variable cannot be given inside
```

```python
def myfunc(**x):
    print(x, type(x))

myfunc(Name = "honey", age = 30, _100 = "hello")  #variables rules should be followed
```

## **Function Overloading**

```python
def myfunc(x):
    return x
def myfunc(x,y):
    return x + y
def myfunc(x,y,z):
    return x + y + z

print(myfunc(100))    # function overwritten
```
