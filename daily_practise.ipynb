37.Write a program using for loop to print all even numbers from 1 to 100.

```python
for n in range(1, 101):
    if n % 2 == 0:
        print(n, end = " ")
```

38.Write a program using for loop to print all odd numbers from 1 to 100. 

```python
for n in range(1,101):
    if n % 2 != 0:
        print(n, end= " ")
```

39.Write a program using for loop to calculate the sum of numbers from 1 to 100. 

```python
sum = 0
for n in range(1,101):
    sum += n 
print(sum)
    
```
