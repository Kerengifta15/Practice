### **Python Standard Functions**

```python
# len()
x = [1, 2, 4, "hi", "hello", True]
y = "hello world"

print(len(y))  # counts letter by letter
print(len(x))  # counts item by item
```

```python
# max()
# min()
# sum()

x = [0,9,5,3,-7,37,100,-12,2]
print(x)

print(max(x))
print(min(x))
print(sum(x))
```

```python
# abs - absolute value

a = 25
a1 = abs(a)
b = -25
b1 = abs(b)
print(a)
print(a1)
print(b)
print(b1)
```

```python
# pow()

print(2**3)  
print(pow(2,3))  # using function
```

```python
# round()
x = 50.55
print(round(x))
```

```python
# slice()

x= [1,9,7,6,4,3,6,2,0]
print(x)

print(x[3:9])
print(x[slice(3,9)]) # using function
```

```python
# sorted()

x = [-9,-5,7,-6,8,3,-1,10,2,-4,-8,0]
print(x)

print(sorted(x))
```

```python
x = [-9,-5,7,-6,8,3,-1,10,2,-4,-8,0]
print(x)

print(sorted(x,reverse = True))  # to print in decending order
```

```python
#chr()
#ord()

print(ord("a"))
print(ord("A"))  # prints ascii values
```

```python
for i in range(33, 127):
    print(chr(i), end= " ") # prints the characters in keyboard
```

### **enumerate()**

```python
# to show all the occurences - enumerate

x = [2,3,6,5,4,7,9,0,3,4,5,1,6,7,3,9,10,2]
print(list(enumerate(x)))
```

```python
x = [2,3,6,5,4,7,9,0,3,4,5,1,6,7,3,9,10,2]
mylist = list(enumerate(x))

for each in mylist:
    print(each)
```

```python
x = [2,3,6,5,4,7,9,0,3,4,5,1,6,7,3,9,10,2]
search = 2
mylist = list(enumerate(x))

for each in mylist:
    if each[1] == search:
        print(f"The element found at the index: {each[0]}")
```

```python
x = "Hello world"
search = "l"
mylist = list(enumerate(x))

for each in mylist:
    if each[1] == search:   #indexes in the list inside the tuple
        print(f"The element found at the index: {each[0]}")   #indexes in the list
```

### **Python modules and Library**

```python
import math

ans = math.sqrt(10)
print(ans)
```

```python
# as - alias

import math as m #to write it in shortcut

ans = m.sqrt(9)
print(ans)
```
