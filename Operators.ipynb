### **Operators**

```python
# Arithmetic Operators
# +, -, *, /, **, //, %

print(15+6)
print(15-6)
print(15*6)
print(15/6)
print(15**6)
print(15//6)
print(15%6)
```

```python
# Assignment Operators
# =, +=, -=, *=, /=, **=, //=, %=

a = 15
print(a)
a += 6
print(a)
a -= 6
print(a)
a *= 6
print(a)
a /= 6
print(a)

b=0
b += 12
print(b)
```

```python
# Comparison Operators
# ==, !=, >, <, >=, <=

print(15==6)
print(15!=6)
print(15>6)
print(15<6)
print(15>=6)
print(15<=6)
```

```python
# Logical Operators
# and, or, not

print("--AND--")

print(10 ==10 and 6>4)
print(10 !=10 and 6>4)
print(10 ==10 and 6<4)
print(10 !=10 and 6<4)

print("--OR--")

print(10 ==10 or 6>4)
print(10 !=10 or 6>4)
print(10 ==10 or 6<4)
print(10 !=10 or 6<4)

print("--NOT--")

print(not 10 ==10)
print(not 10 !=10)
```
