```python
#while loop

row = int(input("Enter number of rows:"))
symbol = input("Enter a symbol:")
a = row
while a>= 1:
    b = 1
    while b<= a:
        print(symbol, end="")
        b += 1
    a-=1
    print() 
```

```python
#while loop
num = int(input("Enter a num:"))
symbol = input("Enter a symbol")

a = 1
while a <= num:
    b = 1
    while b <= num - a:      #space printing
        print(" ",end="")
        b += 1
    b = 1
    while b <= a:
        print(symbol, end="")
        b += 1
    a += 1
    print()
```
