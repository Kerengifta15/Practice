**34.Write a program using while loop to reverse a given number.** 

```python
num = int(input("Enter a num: "))
print("Given number:", num)
rev = 0
while num > 0:
    a = num % 10
    rev = rev * 10 + a
    num //= 10
print("Reversed number:",rev)
```

**35.Write a program using while loop to check whether a number is a palindrome.** 

```python
num = int(input("Enter a num:"))
print("Number:",num)
original = num
rev = 0
while num > 0:
    a = num % 10
    rev = rev * 10 + a
    num //= 10
if original == rev:
    print("It is a palidrome")
else:
    print("Not a palindrome")
```
