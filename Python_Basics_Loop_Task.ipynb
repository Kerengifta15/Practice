3.Write a program to input the length and breadth of a rectangle and calculate its area and 
perimeter. 

```python
length = int(input("Enter the length of the rectangle: "))
breadth = int(input("Enter the width of the rectangle: "))
area = length * breadth
perimeter = 2 * (length + breadth)
print("Area:", area)
print("Perimeter:", perimeter)
```

8.Write a program to calculate simple interest using principal, rate, and time.

9.Write a program to demonstrate all arithmetic operators using two numbers. 

```python
# +, -, *, /, **, //, %

a = int(input("Enter a number: "))
b = int(input("Enter a number: "))
add = a + b
sub = a - b
mul = a * b
div = a / b
mod = a % b
expo = a ** b
floor_div = a // b
print("Addition:", add)
print("Subtraction:", sub) 
print("Multiplication:", mul)
print("Division:", div)
print("Modulus:", mod)
print("Exponentiation:", expo)
print("Floor Division:", floor_div)
```

12.Write a program to check whether a number is divisible by both 3 and 5 using logical 
operators. 

```python
# and, or, not

a = 13

if (a%3==0) & (a%5==0):
    print(f'{a} is divisible by both 3 and 5')
else:
    print(f"{a} is not divisible by both 3 and 5")
```
