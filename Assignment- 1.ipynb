**3. Explain type casting with two examples.**  
Type casting means converting a value from one data type to another.

```python
# e.g.  
x = int("10")
y = float(10)
z = bool(10)
print(x, type(x))
print(y, type(y))
print(z, type(z))  
```

**4. What is the difference between a list and a tuple?**  
The main difference is that a list is mutable, while a tuple is immutable.

```python
# e.g.
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
print(my_list, type(my_list))
print(my_tuple, type(my_tuple))
```

**5. What is the difference between a set and a dictionary?**  
**Set:**  
Unordered and unindexed.  
No duplicate members.  
Denoted by {}.<br>  
**Dictionary:**   
Unordered, changeable and indexed.  
No duplicate keys.   
Denoted by {key:value}.

**2. Simple Calculator**

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print("First number:", num1)
print("Second number:", num2)
Sum = num1 + num2
print("Sum:", Sum)
Sub = num1 - num2
print("Subtraction:", Sub)
Mul = num1 * num2
print("Multiplication:", Mul)
Div = num1 / num2
print("Division:", Div)
Floor_Div = num1 // num2
print("Floor Division:", Floor_Div)
Mod = num1 % num2
print("Modulus:", Mod)
```
