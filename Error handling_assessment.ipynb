### **Python Error Handling - Assessment**

**1. What is the difference between syntax errors and runtime errors in python ?**    
           
**Syntax Error:**    
A syntax error occurs when the code is written incorrectly according to Python's rules. The program will not run until the error is fixed.    

```python
#Example:
if age > 18
    print("Adult")
```

**Runtime Error:**    
A runtime error occurs while the program is running. The code is written correctly, but something goes wrong during execution.

```python
#Example:
a = 10
b = 0
print(a / b)
```

**2.Explain the role of try, except, else and finally blocks.**
     
• *try block* - lets you test a block of code for errors.   
• *except block* - lets you handle the error.   
• *finally block* - lets you execute code, regardless of the result of the  try and except blocks.    
• *else block* - works only if try is correct.       

```python
#Example:
try:
    a = 10
    b = 2
    print(a / b)
except:
    print("An error occurred")
else:
    print("No error")
finally:
    print("Program completed")
```

**3. What happens if an exception is raised but not handled ?**   

If an exception is raised but not handled, the program stops running and Python displays an error message.     

```python
#Example:
a = 10
b = 0
print(a / b)
```

**4. What is the difference between Exception and Base Exception in Python ?**

• *BaseException* – It is the main/base class for all built-in exceptions in Python.       
• *Exception* – It is a child class of BaseException and is used for most normal errors that we handle in programs.    


```python
#Example:
try:
    print(10 / 0)
except Exception:
    print("An error occurred")
# Here, ZeroDivisionError is a child of Exception.
```

**5. Explain the use of raise keyword with an example.**

The raise keyword is used to create an error manually when a specific condition is not met.   

```python
age = int(input("Enter your age: "))
print("Age:", age)

if age < 18:
    raise Exception("Age must be 18 or above")
print("You can vote")
```

**6. How can you create a custom exception class in python ?**

A custom exception is an error that we create ourselves for a specific situation.   
We can create it by making a class that inherits from Exception.    

```python
class AgeError(Exception):
    pass
age = 15
if age < 18:
    raise AgeError("Age must be 18 or above")   #Ageerror is a custom exception
```

**7. What is the purpose of the assert statement ,and how does it differ from raising an exception ?**

**Purpose of assert statement:**   
The assert statement is used to check whether a condition is true.

•  If the condition is True, the program continues.   
•  If the condition is False, Python raises an AssertionError.

```python
# Example:
age = 20
assert age >= 18
print("You can vote") # Here, the condition is true, so the program continues.
```

**Difference from raise**   
*assert* → Used to check a condition.   
*raise* → Used to manually create an exception.   

```python
age = 15
assert age >= 18 # the condition is false, Python gives an AssertionError.
```

```python
age = 15
if age < 18:
    raise Exception("Age must be 18 or above") # Here, we manually raise an exception.
```

**8. Can we use multiple except blocks in python ? Give an example.**   

Yes, we can use multiple except blocks in Python to handle different types of errors separately.

```python
#Example:
try:
    a = int(input("Enter a number: "))
    print(10 / a)
except ValueError:
    print("Please enter a number")
except ZeroDivisionError:
    print("Cannot divide by zero")

#Here:
# ValueError handles invalid input, like entering "abc".
# ZeroDivisionError handles division by zero.
```

**9. What happens if you use a finally block without an except block ?**

Yes, we can use a finally block without an except block.   
The finally block will always run, whether an error occurs or not.   

```python
try:
    print(10 / 2)
finally:
    print("Program completed")
```

**10. What is the difference between checked and unchecked exceptions ? Does python support checked exceptions ?**

**Checked exceptions** are errors that the programmer is required to handle before running the program.   
**Unchecked exceptions** are errors that the programmer is not required to handle. They occur when the program runs.   

No, Python does not have checked exceptions.  
Python uses unchecked exceptions, so you can choose whether to handle an exception using try and except.   

```python
#Example:
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero")
```
