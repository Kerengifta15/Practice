```python
#0   1   3   4   5   6   7   8   9 Positive index   

a = "Helloworld"

print(a)
print(a[1])
print(a[-2])
print(a[0:9])
print(a[0:-1])
print(a[5:])
print(a[:5])
print(a[:])
print(a[0:10:2])
print(a[::5])
print(a[::-1])  #to be remember - used frequently
```

## **String Functions**

```python
a = "helloworld"
print(a)
```

```python
a = "hello1234 world"
x = a.isalpha()   # checks whether all in a is a alphabet
print(x)
```

```python
a = "helloworld"    # does not store in memory
print(a.isalpha())   

print("keren".isalpha())   # does not store in memory
```

```python
a = "helloworld "   # space is a special character
x = a.isalpha()
print(x)
```

```python
a = "12345678"
x = a.isdigit()
print(x)
```

```python
a = "1234helloworld"
x = a.isalnum()
print(x)
```

```python
a = "1234 helloworld"  #space
x = a.isalnum()
print(x)
```

```python
# islower

a = "helloworld"
x = a.islower()
print(x)

a = "hello world"   #does not consider space
x = a.islower()
print(x)

a = "hello 134world" # Special characters and numbers can also be given
x = a.islower()
print(x)
```

```python
a = "HELLOWORLD"
x = a.isupper()
print(x)
```

```python
#upper
#converts the string into upper case

a = "123Hello World"

x = a.upper()
print(x)
```

```python
#lower

a = "HELLO WORLD"
x = a.lower()
print(x)
```

```python
# ends with()

a = "ABCD Keren"
x = a.endswith("N")
print(x)

a = "ABCD Keren"
x = a.endswith("n")
print(x)

a = "ABCD Keren.txt"
x = a.endswith("txt")
print(x)
```

```python
mylist =["new.txt", "new.py", "hello.py"]

for each in mylist:
    if each.endswith(".py"):
        print(each)
```

```python
mylist =["new.txt", "new.py", "hello.py"]

for each in mylist:
    if each.startswith("n"):
        print(each)
```
