### **String Formatting**

* %d - integer
* %s - string
* %f - float  with default 6 decimal places
* %.2f - float  with 2 decimal places
* %.10f- float  with 10 decimal places

```python
name = "John"
age = 28
height = 160.15

print("%s is %d year old and height is %.2f" %(name,age,height))
```

```python
length = 123.5050
width = 150.5000
height = 7050.1500

print("length = %f, width =")
```

```python
#f-format

quantity = 3
itemno = 500
price = 50

print(f"I want {quantity} pieces of item number {itemno} for $ {price} dollar")
```
