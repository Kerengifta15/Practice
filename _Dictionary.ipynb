### **Dictionary Functions**

```python
mydict = {
    "Name" : "Keren",
    "Age"  : 22,
    "City" : "Chennai",
    500    : 50.55
}

print(mydict,type(mydict))
print(mydict[500])
mydict["City"] = "Mumbai"   #add element
print(mydict)
```

```python
#keys()

mydict = {
    "Name" : "Keren",
    "Age"  : 22,
    "City" : "Chennai",
    500    : 50.55
}
print(mydict)
print()
print(list(mydict.keys()))
print()

ans = list(mydict.keys())
print(ans[1])
```

```python
#value()

mydict = {
    "Name" : "Keren",
    "Age"  : 22,
    "City" : "Chennai",
    500    : 50.55
}
print(mydict)
print(mydict.values())
print(list(mydict.values()))

for each in mydict.values():
    print(each)
```

```python
#items()

mydict = {
    "Name" : "Keren",
    "Age"  : 22,
    "City" : "Chennai",
    500    : 50.55
}
print(mydict)
print()

print(mydict.items())
print()

print(list(mydict.items()))
print()

for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

```python
#pop

#list = pop(),pop(index)

x =[100, 30.55, "hello", 100, "hi", 30.55]
x.pop()
print(x)     # remove last element

x.pop(2)
print(x)

#set = pop()

y ={100, 20.55, 100, "hi",10, 20.55}
y.pop()
print(y)
```

```python
#dict - pop(keys)

mydict = {
    "Name" : "Keren",
    "Age"  : 22,
    "City" : "Chennai",
    500    : 50.55
}
print(mydict)

mydict.pop("Age")
print(mydict)
```

```python
#popitem()

mydict = {
    "Name" : "Keren",
    "Age"  : 22,
    "City" : "Chennai",
    500    : 50.55
}
print(mydict)

mydict.popitem() # removes last inserted key value pair

```
