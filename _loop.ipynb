### **Python while loop** (Condition statements)

*Syntax*
* **while \<condtion\>:**

```python
a = 1  #initialization/start
while a <= 10:  #condition/end
    print(a, end=" ")
    a += 1   #increment/jump

print("\nloop over")
```

```python
a = 100  #initialization/start
while a >=0:  #condition/end
    print(a, end=" ")
    a -= 10   #increment/jump

print("\nloop over")
```

**break** -- stop the loop

```python
a = 1
while a > 0:
    if a ==5:
        print(a)
        break
    print(a, end=" ")
    a += 1
```

**continue** - go back to the top of the loop

```python
a = 1  
while a <= 20:  
    if a % 3 == 0:
        a += 1
        continue
    print(a, end=" ")
    a += 1
print("\nloop over")
   
```

**Nested While Loop**

```python
a = 1 
while a <= 3:
    b = 1
    while b <= 3:
        print(b, end="")
        b += 1
    a += 1
    print()
```

```python
n= 3
while n:
    print(n)
    n -= 1  
```

```python
i = 1
while i <= 3:
    print("Hello")
    i += 1
else:
    print("Done")
```

## **Python For Loop**
Syntax  
**for \<variable_name\> in \<str/list/tuple/set/dict\>:**

```python
mystr = "Hello world"

for x in mystr:  #fetches letter by letter
    print(x, end="")
```

```python
mylist = [10,20.55,"hello",True,20.55,"hello"]
for x in mylist: 
    print(x)
```

**range()** - is a python standard function used for numbers

```python
for x in range(10+1):  #End before
    print(x)
```

```python
for x in range(5,10+1):  #Start with end before
    print(x, end=" ")
```

```python
num = int(input("Enter a number: "))
for x in range(50,num+1):
    print(x, end=" ")
```

```python
for x in range(0, 10+1, 2):   #jump
    print(x, end=" ")  
```

```python
for x in range(10,0,-1):  #fetches number from 10 to 1
    print(x, end=" ")
```

### **for with else**

```python
# if loop completes successfully, else block will be executed
# if loop is unsuccessful, else block will not be executed

for x in "hello world":
    print(x, end="")
else:
    print("\nThe above loop has completed successfully")
```

```python
for x in "hello world":
    if x == " ":
        break
    print(x, end="")
else:
    print("\nThe above loop has completed successfully")
```

```python
for x in "helloworld":
    if x == " ":
        break
    print(x, end="")
else:
    print("\nThe above loop has completed successfully")
```
