## **if, elif, else rules:**
**if -> only once per code block(mandatory)**  
**elif -> as many times per code block (optional)**  
**else -> only once per code block(optional)**

## **Short hand if and else**

```python
a = 100
b = 500
if a > b: print("a is greater than b")   
else: print("b is greater than a")
```

```python
a = 1000
b = 500
print("a is greater than b") if a > b else print("b is greater than a")

# colon(:) should not be used in one line if else statement. It will give syntax error.
```

```python
# Game 1

score = 100
secret_number = 50
guess = int(input("Enter your guess: "))

if guess == secret_number:
    print("Correct")
    score += 20
    print("Your score is:", score)

elif guess > secret_number:
    print("Too high")
    score -= 10
    print("Your score is:", score)

else:
    print("Too low")
    score -= 10
    print("Your score is:", score)
```
